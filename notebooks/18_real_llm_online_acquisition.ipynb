# P2.3 Real LLM Online Acquisition

This notebook runs the P2.1 online loop with a real OpenAI-compatible language model as the only teacher. It records per-episode LLM usage so cumulative baseline and Paradigm token curves, time-to-reflex, and old-family retention can be inspected.


```python
import json
from pathlib import Path

from paradigm.llm_controller import OpenAICompatibleCodingDeliberator
from paradigm.p23 import run_p23_live_online_benchmark
```

Required environment variables: `PARADIGM_LLM_BASE_URL` and `PARADIGM_LLM_MODEL`. The recorded run in `results/core_p23/` used a local `qwen3:4b-instruct` served by Ollama. Set `quick=False` to reproduce the full 69-episode stream.


```python
controller = OpenAICompatibleCodingDeliberator.from_env()
result = run_p23_live_online_benchmark(controller, quick=True)
result['novel_family']
```

The recorded full-run artifacts can be loaded without a model endpoint.


```python
recorded = json.loads(Path('../results/core_p23/core_p23_online_llm.json').read_text())
for row in recorded['per_episode'][::6]:
    print(row['episode'], row['phase'], row['family'], row['cumulative_tokens_baseline'], row['cumulative_tokens_online'], row['cumulative_tokens_saved'])
```
