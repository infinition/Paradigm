# Paradigm World

Research bridge to [FluidWorld](https://github.com/infinition/FluidWorld) and [arXiv:2603.21315](https://arxiv.org/abs/2603.21315).

The first question is narrow: does predictive latent state improve the decision to execute a reflex or defer to deliberation?

## Proposed interface

```text
observation -> external world-state encoder -> latent state
                                      |
                                      v
                            Paradigm reflex gate
                                      |
                         accept ------+------ defer
```

A later stage can evaluate short candidate-action rollouts, but Paradigm Core should remain usable without FluidWorld.

````python
from dataclasses import dataclass
import numpy as np

@dataclass
class PredictiveState:
    current: np.ndarray
    predicted_next: np.ndarray
    uncertainty: float

def should_defer(state: PredictiveState, threshold: float = 0.25) -> bool:
    return state.uncertainty > threshold

print("Interface scaffold ready")
````
