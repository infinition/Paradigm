# Bounded plasticity

Research bridge to [drift-contract](https://github.com/infinition/drift-contract) and the [public preprint](https://infinition.github.io/infinition/assets/papers/drift-bounded-spectral-updates.pdf).

The source work studies deep local learning. Paradigm transfers only the matrix update mechanics in this first benchmark.

```
from pathlib import Path
import json
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
result = json.loads((ROOT / 'results/core_p10/core_p10_bounded_plasticity.json').read_text())
```

## Distribution shock

Five seeds compare Adam, Adam with replay, a faster drift budget, and a conservative drift budget.

```
{method: {metric: values['mean'] for metric, values in metrics.items() if isinstance(values, dict) and 'mean' in values} for method, metrics in result['distribution_shock']['methods'].items()}
```

## Interpretation

The budget exposes a stability versus adaptation-speed tradeoff. The benchmark does not transfer the source paper's local-learning accuracy or depth claims to Paradigm.
