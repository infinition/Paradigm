# Core P1.1: multi-layer continual adaptation

This notebook inspects the recorded P1.1 result. The benchmark compares continual adaptation methods and tests explicit neural parameter-delta trust.


```python
import json
from pathlib import Path

result = json.loads(Path('../results/core_p11/core_p11_multilayer.json').read_text())
list(result['continual_adaptation'])
```

```python
rows = []
for name, item in result['continual_adaptation'].items():
    rows.append({
        'method': name,
        'current': item['final_current_accuracy']['mean'],
        'mean_seen': item['final_mean_seen_accuracy']['mean'],
        'worst_seen': item['final_worst_seen_accuracy']['mean'],
        'forgetting': item['final_mean_forgetting']['mean'],
        'seconds': item['total_adapt_seconds']['mean'],
        'step_drift': None if item['mean_max_step_drift'] is None else item['mean_max_step_drift']['mean'],
    })
rows
```
