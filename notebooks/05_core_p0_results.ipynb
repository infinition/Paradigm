# Core P0 results

This notebook reads the reproducible Core P0 result file and summarizes the selected reflex for each scenario.

```python
import json
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
results = json.loads((ROOT / "results/core_p0/core_p0_suite.json").read_text())
rows = []
for scenario, payload in results["scenarios"].items():
    selected = payload["selected_reflex"]
    metrics = payload["methods"]["selected_reflex"]
    rows.append({
        "scenario": scenario,
        "backend": selected["backend"],
        "coverage": metrics["coverage"],
        "selective_accuracy": metrics["selective_accuracy"],
        "ece": metrics["ece"],
        "latency_us_per_item": metrics["latency_us_per_item"],
        "reference_accuracy": payload["reference_accuracy"],
        "quality_pass": payload["go_no_go"]["quality_pass"],
    })
df = pd.DataFrame(rows)
df
```

## Interpretation

The selector should prefer the least complex backend that satisfies the declared constraints. The current synthetic result supports that rule, but it does not establish a real efficiency gain against an expensive deliberative system.
