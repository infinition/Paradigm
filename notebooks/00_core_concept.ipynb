# Paradigm Core: controlled decision traces

This notebook creates the first controlled dataset for reflex compilation. The reference policy is deterministic so every later failure is attributable to the compiler, calibration, or gating rather than label ambiguity.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT / "src"))

import numpy as np
from paradigm.synthetic import FEATURE_NAMES, make_traces

traces = make_traces(4000, seed=0)
X = np.stack([t.features for t in traces])
y = np.array([t.action for t in traces])
print("shape:", X.shape)
print("features:", FEATURE_NAMES)
for label in sorted(set(y)):
    print(label, int((y == label).sum()))
```
