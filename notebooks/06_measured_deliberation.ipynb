# Core P0.1: Measured Deliberation

Loads the recorded benchmark that replaces the assumed cost ratio with a measured local deliberative surrogate. The surrogate is intentionally not presented as an LLM or planner.

```
from pathlib import Path
import json
ROOT = Path('..') if Path.cwd().name == 'notebooks' else Path('.')
DATA = json.loads((ROOT / 'results/core_p01/core_p01_measured.json').read_text())
DATA.keys()
```

```
import pandas as pd
rows=[]
for name, item in DATA['scenarios'].items():
    row={'scenario': name, **item['metrics']}
    rows.append(row)
pd.DataFrame(rows)[['scenario','selected_backend','coverage','selective_accuracy','deliberative_accuracy','system_accuracy','speedup_vs_deliberation','fallback_rate']]
```

The noisy scenario is intentionally retained even when it has a smaller efficiency margin. Backend complexity is part of the research question.
