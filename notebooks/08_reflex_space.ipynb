## Interpretation

- Reflex-level signatures expose candidate corruption that input-space gates cannot see.
- Weak-pool rejection remains a coverage problem, not evidence of invalidity.
- Parameter and behavior views are complementary.
- Promotion must still include explicit task, subgroup, calibration, and drift checks.
